# MiniLearn: Final Project Report

## 1. Introduction and Project Context

MiniLearn is a minimal, educational machine learning library developed from scratch as part of the CSE 432/532 Speech Emotion Recognition (SER) project. The goal was to gain a deep understanding of core machine learning algorithms, model evaluation, and the end-to-end workflow of a real-world ML project by re-implementing key components in a scikit-learn-like API. The library supports both classical and neural network models, and is used throughout the project’s Jupyter notebooks for experimentation, benchmarking, and reporting.

---

## 2. High-Level Architecture and Design Principles

MiniLearn is organized as a modular Python package, inspired by scikit-learn’s structure. The main submodules are:
- `classifiers`: Classical supervised learning models (logistic regression, k-NN, naive Bayes, decision tree, SVM)
- `neural_network`: Simple feedforward artificial neural network (ANN) and experiment helpers
- `preprocessing`: Data scaling, splitting, and dimensionality reduction (PCA)
- `metrics`: Evaluation metrics for classification and clustering
- `model_selection`: Cross-validation utilities
- `unsupervised`: K-Means clustering

**Design Principles:**
- Consistent, minimal API: All models implement `fit`, `predict`, and `score` methods.
- Educational clarity: Code is well-documented, readable, and avoids unnecessary abstraction.
- Reproducibility: Random seeds and stratified splits are supported throughout.
- Compatibility: Follows scikit-learn conventions for easy integration and comparison.

---

## 3. Module-by-Module Overview

### 3.1 Classifiers (`minilearn.classifiers`)

Implements core supervised learning algorithms, each as a separate class:
- **LogisticRegression**: Multiclass softmax regression with batch gradient descent and L2 regularization.
- **KNearestNeighbors**: k-NN classifier with uniform/distance weighting and Euclidean distance.
- **GaussianNaiveBayes**: Gaussian Naive Bayes with per-class mean/variance estimation.
- **DecisionTreeClassifier**: CART-style binary tree with Gini impurity, continuous splits, and depth/leaf controls.
- **SupportVectorMachine**: Linear SVM using hinge loss, gradient descent, and One-vs-Rest multiclass strategy.

All classifiers implement `fit(X, y)`, `predict(X)`, and `score(X, y)` methods, following scikit-learn conventions. Each is designed for clarity and educational value, with detailed docstrings and parameter validation.

---

### 3.2 Neural Network (`minilearn.neural_network`)

- **SimpleANNClassifier**: Implements a basic feedforward neural network with configurable hidden layers, activation functions (ReLU, tanh, sigmoid), batch training, L2 regularization, and early stopping. Designed for multiclass classification with softmax output.
- **Experiment Helpers**: `tune_simple_ann` automates hyperparameter grid search and evaluation; `load_dataset` standardizes data loading for experiments.

The ANN module enables direct comparison with classical models and TensorFlow/Keras implementations, supporting educational benchmarking.

---

### 3.3 Preprocessing (`minilearn.preprocessing`)

- **StandardScaler**: Standardizes features by removing the mean and scaling to unit variance.
- **train_test_split**: Splits data into train/test sets, with optional stratification to preserve class balance.
- **PCA**: Principal Component Analysis for dimensionality reduction, implemented via SVD.

All utilities are compatible with scikit-learn’s API, supporting `fit`, `transform`, and `fit_transform` methods where appropriate.

---

### 3.4 Metrics (`minilearn.metrics`)

- **Classification Metrics**: `accuracy_score`, `precision_score`, `recall_score`, `f1_score` (macro/micro/weighted averaging)
- **Confusion Matrix**: `confusion_matrix` for evaluating classifier predictions
- **Clustering Metrics**: `adjusted_rand_index`, `normalized_mutual_information` for unsupervised evaluation

All metrics are implemented from first principles using NumPy, with clear docstrings and scikit-learn-style signatures.

---

### 3.5 Model Selection (`minilearn.model_selection`)

- **StratifiedKFold**: Cross-validation splitter that preserves class distribution across folds, supporting reproducible shuffling and API compatibility with scikit-learn.
- **cross_validate**: Utility to evaluate models using cross-validation, returning train/test scores for each fold.

These tools enable robust model evaluation and hyperparameter tuning, critical for fair comparison and reproducibility.

---

### 3.6 Unsupervised Learning (`minilearn.unsupervised`)

- **KMeans**: Implements the K-Means clustering algorithm with random initialization, iterative centroid updates, inertia tracking, and prediction. API mirrors scikit-learn’s `fit`, `predict`, and `fit_predict` methods.

This module supports unsupervised learning and clustering evaluation, complementing the classification focus of the library.

---

## 4. Notable Design Decisions and Lessons Learned

- **API Consistency**: All models and utilities follow a unified API, making it easy to swap components and compare results.
- **Educational Focus**: Simplicity and clarity were prioritized over performance or advanced features, to maximize learning value.
- **Reproducibility**: Random seeds, stratified splits, and deterministic algorithms ensure results can be replicated.
- **Error Handling**: Extensive parameter validation and informative error messages aid debugging and user experience.
- **Comparison with Industry Tools**: The library was designed to be interoperable with scikit-learn and TensorFlow/Keras, enabling direct benchmarking and validation.
